# Calcs & Assumptions Mining (sheet)

```python
# Mining estimates
labor_cost = 1.8                                # $/ton ore, Calcs & Assumptions Mining'!B6
diesel_cost = 1.2                               # $/ton ore, Calcs & Assumptions Mining'!B7
electricity_cost = 0.3                          # $/ton ore, Calcs & Assumptions Mining'!B8
maintenance_cost = 1.2                          # $/ton ore, Calcs & Assumptions Mining'!B9
drill_blast_consumables_cost = 0.6              # $/ton ore, Calcs & Assumptions Mining'!B10
general_mine_overhead_cost = 0.8                # $/ton ore, Calcs & Assumptions Mining'!B11
total_cost = (labor_cost+                       # $/ton ore, Calcs & Assumptions Mining'!B12
              diesel_cost+
              electricity_cost+
              maintenance_cost+
              drill_blast_consumables_cost+
              general_mine_overhead_cost)

mining_percent_of_electricity = (electricity_cost/total_cost)*100   # % of electricity used by mining, Calcs & Assumptions Mining'!B14
estimate_of_mining_vs_crushing = 0.45   # Calcs & Assumptions Mining'!D29
```

```python
## Mining
# Electricity (kWh) and Electricity Usage/LT Pellets (kWh/LTP)
cost_per_LTP = 15.3                         # $/LTP, 'Calcs & Assumptions Mining'!B19
LTP = np.mean([7580920,7631260,7631980,7650141,7678514,7708582,6323241,7457805])  # LTP, 'Calcs & Assumptions Mining'!B20

LTP_cost = cost_per_LTP * LTP               # $ USD for LTP, 'Calcs & Assumptions Mining'!B22
cost_per_kWh = 0.0647                       # $/kWh, 'Calcs & Assumptions Mining'!B23
total_kWh = LTP_cost / cost_per_kWh         # kWh, 'Calcs & Assumptions Mining'!B24
kWh_per_LTP = cost_per_LTP/ cost_per_kWh    # kWh/LTP, 'Calcs & Assumptions Mining'!B25

mine_electricity_consumption = total_kWh * (mining_percent_of_electricity/100) # kWh, 'Calcs & Assumptions Mining'!B27 (original value 89,674,185.082)
mine_electricity_consumption_per_LTP = kWh_per_LTP * (mining_percent_of_electricity/100)     # kWh/LTP, 'Calcs & Assumptions Mining'!B28 (original value 12.0242056)

print("Tilden Mining Electricity (kWh)", mine_electricity_consumption)
print("Tilden Mining Electricity (kWh/LTP)", mine_electricity_consumption_per_LTP)

# Fuel Consumption (Diesel + Gasoline) (gal)
num_haul_trucks = 17                                    # number of units, 'Calcs & Assumptions Mining'!B33
num_shovels = 5                                         # number of units, 'Calcs & Assumptions Mining'!B34
num_dozers = 7                                          # number of units, 'Calcs & Assumptions Mining'!B35
num_loaders = 4                                         # number of units, 'Calcs & Assumptions Mining'!B36
num_graders = 4                                         # number of units, 'Calcs & Assumptions Mining'!B37
num_drills = 4                                          # number of units, 'Calcs & Assumptions Mining'!B38
num_service_support = 1                                 # number of units, 'Calcs & Assumptions Mining'!B39
MLT_total_material = np.mean([57.7,62,62,61.8,58.3])    # LT/year, 'Calcs & Assumptions Mining'!B40
Hibtac_MLT_total_material = np.mean([49.6,63,66.3,62.8,63.7,61.9,63.9,64.8,64.7,60,45.2,59.7]) # Unsure, 'Calcs & Assumptions Mining'!C40
Hibtac_gasoline_consumption = np.mean([101158,106146,82833]) # gal/year, 'Calcs & Assumptions Mining'!C42 (original value = 96,712)
Hibtac_diesel_consumption = np.mean([7135684,7157733,5317569]) # gal/year, 'Calcs & Assumptions Mining'!C43 (original value = 6,536,995)


gasoline_consumption = (Hibtac_gasoline_consumption/Hibtac_MLT_total_material) * MLT_total_material     # gal/year, 'Calcs & Assumptions Mining'!B42 (original value = 96,542)

haul_truck_diesel_consumption = (((((Hibtac_diesel_consumption * 0.73)/ 1E6)/Hibtac_MLT_total_material)/14)*num_haul_trucks*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B48
shovel_diesel_consumption = (((((Hibtac_diesel_consumption * 0.08)/ 1E6)/Hibtac_MLT_total_material)/5)*num_shovels*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B49
dozer_diesel_consumption = (((((Hibtac_diesel_consumption * 0.04)/ 1E6)/Hibtac_MLT_total_material)/6)*num_dozers*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B50  
loader_diesel_consumption = (((((Hibtac_diesel_consumption * 0.06)/ 1E6)/Hibtac_MLT_total_material)/1)*num_loaders*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B51                    
grader_diesel_consumption = (((((Hibtac_diesel_consumption * 0.02)/ 1E6)/Hibtac_MLT_total_material)/2)*num_graders*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B52
drill_diesel_consumption = (((((Hibtac_diesel_consumption * 0.02)/ 1E6)/Hibtac_MLT_total_material)/4)*num_drills*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B53  
service_diesel_consumption = (((((Hibtac_diesel_consumption * 0.05)/ 1E6)/Hibtac_MLT_total_material)/1)*num_service_support*MLT_total_material*1E6) # gal, 'Calcs & Assumptions Mining'!B54
diesel_consumption = (haul_truck_diesel_consumption+         # gal/year, 'Calcs & Assumptions Mining'!B55 (original value = 8,894,828)
                     shovel_diesel_consumption+
                     dozer_diesel_consumption+
                     loader_diesel_consumption+
                     grader_diesel_consumption+
                     drill_diesel_consumption+
                     service_diesel_consumption)

total_fuel_consumption = gasoline_consumption + diesel_consumption # gal/year, 'Calcs & Assumptions Mining'!B44 (original value 8,991,370.221)

print("Tilden Mining Fuel Consumption (Diesel + Gasoline, gal)", total_fuel_consumption)

# Material Movement (WLT/year)
total_material_movement = MLT_total_material*1E6    # LT/year, 'Calcs & Assumptions Mining'!B68
print("Tilden Mining Material Movement (WLT/yr)", total_material_movement)

#Overburden (Stripping Ratio)
stripping_ratio = 1.2       # ratio, 'Calcs & Assumptions Mining'!B67
print("Tilden Mining Overburden (Stripping Ratio)", stripping_ratio)

# MagFe (%)
MagFe_percent = 34.7        # % Fe, hardcoded 
print("Tilden Mining MagFe (%)", MagFe_percent)

# Mineral Composition, 'Calcs & Assumptions Mining'!B81
percent_magnetite = [5,15]  # 5-15%
percent_hematite = [35,60]  # 35-60%
percent_silica = [30,45]    # 30-45%
percent_silicate = [5,15]   # 5-15%
percent_carbonates = [0,5]  # 0-5%
mineral_composition = {
    "Magnetite": percent_magnetite,
    "Hematite": percent_hematite,
    "Silica": percent_silica,
    "Silicates": percent_silicate,
    "Carbonates": percent_carbonates
}

print("Tilden Mining Mineral Composition (ranges)", mineral_composition)

# Product Size
p80_size = [18,24]  # 18-24 inches
f100_size = [36,48] # 36-48 inches

print("Tilden Mining P80 size (inches, range)", p80_size)
print("Tilden Mining F100 size (inches, range)", f100_size)

# Crude Oil Reserve (MLT)
crude_oil_reserve_MLT = 520.0
print("Tilden Mining Crude Ore Reserve (MLT)", crude_oil_reserve_MLT)
```
